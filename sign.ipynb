```python
import pandas as pd
from IPython.display import display
import matplotlib.pyplot as plt
from PIL import Image

# Load results csv
results = pd.read_csv('runs/train/exp23/results.csv')

# Strip whitespace from column names
results.columns = results.columns.str.strip()

# Display metrics table
display(results)

# Plot metrics
results[['metrics/precision',
         'metrics/recall',
         'metrics/mAP_0.5',
         'metrics/mAP_0.5:0.95']].plot(figsize=(10,5))

plt.title("YOLOv5 Evaluation Metrics")
plt.show()
```

```python
import gradio as gr
import torch

# Load trained model
model = torch.hub.load(
    'ultralytics/yolov5',
    'custom',
    path='runs/train/exp22/weights/best.pt', # Corrected path from exp35 to exp22
    force_reload=True
)

def detect_image(image):

    # Perform detection
    results = model(image)

    # Render detected image
    results.render()

    # Return detected image
    return results.ims[0]

demo = gr.Interface(
    fn=detect_image,
    inputs=gr.Image(type="pil"),
    outputs=gr.Image(type="numpy"),
    title="Sign Language Detection"
)

demo.launch(share=True)
```
